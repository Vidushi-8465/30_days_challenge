```python
import pandas as pd                                               #Pandas to handle the data
from sklearn.model_selection import train_test_split              #To Split the data for training and testing
from sklearn.feature_extraction.text import TfidfVectorizer       # To extract the data using TF-IDF
from sklearn.naive_bayes import MultinomialNB                     #naive_bayes Algorithm
from sklearn.metrics import accuracy_score, classification_report # The calculations

# 1. Load dataset
# we are loading the dataset i.e spam.csv
df = pd.read_csv("spam.csv", encoding="latin-1")
df = df[['v1', 'v2']]  # V1, V2 are columns i.e v1= label , v2=text
df.columns = ['label', 'text']

# 2. Convert labels to numeric (as  ml model reads numbers and not text labels)
df['label_num'] = df['label'].map({'ham': 0, 'spam': 1})

# 3. Split data
X_train, X_test, y_train, y_test = train_test_split(
    df['text'], 
    df['label_num'], 
    test_size=0.2, # Training the data= 80%, Testing the Data = 20%
    random_state=42 # This ensures that the result is consistent everytime the code is running
)

# 4. Vectorize text (i.e to use TF- IDF)
tfidf = TfidfVectorizer(stop_words='english') # this removes english stop words like a ,the ,or ,and etc
X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

# 5. Train naive bayes model of Tf-IDf feature
model = MultinomialNB()
model.fit(X_train_tfidf, y_train)

# 6. Predict
y_pred = model.predict(X_test_tfidf)

# 7. Evaluate
print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))

# 8. Test with custom input
while True:
    msg = input("Enter a message: ")
    if msg.lower() == "exit":
        break
    msg_tfidf = tfidf.transform([msg])
    pred = model.predict(msg_tfidf)
    print("Spam" if pred == 1 else "Ham")
```
